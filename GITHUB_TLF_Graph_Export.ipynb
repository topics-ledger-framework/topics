## 🌌 Data Archival: Exporting the Simulation Framework

This execution cell serves as the primary data extraction and archival utility for the simulation. Rather than solely generating a 2D visual render, this script interfaces with Google Cloud to package the underlying mathematical architecture of the emergent system into a highly portable, standardized dataset.

By executing this cell, the simulation state is separated into two distinct, explorable data structures: the causal network (graph data) and the foundational probability space (tensor data).

### 1. The Causal Directed Acyclic Graph (DAG) — "Luminous Matter"
This dataset represents the emergent, causal topology of the simulation—the discrete computational events, connection pathways, and supermassive hubs of algorithmic certainty.

* **Generated Files:** `tlf_universe_nodes.csv` and `tlf_universe_edges.csv`
* **Output Specifications:**
  * **Nodes CSV:** Contains the metadata for every computational event. Columns include a unique identifier (`id`), chronological birth step (`step`), localized Shannon entropy (`entropy`), accrued gravitational mass (`mass`), and precise 3D spatial coordinates (`x`, `y`, `z`).
  * **Edges CSV:** Contains the directional vectors (`source`, `target`) mapping the flow of causality and probabilistic accretion between nodes.
* **Primary Use Case:** These lightweight, tabular files are optimized for immediate ingestion into network analysis and visualization environments. Researchers can load these directly into **R (`igraph`)**, **Python (`NetworkX` or `Plotly`)**, or **Gephi** to quantitatively analyze the structural evolution and clustering coefficients of the network.

### 2. The Probability Amplitude Space — "Dark Matter"
This dataset archives the underlying, massive computational state space from which the causal structures emerged. For a 16-qubit superposition spanning approximately 200 algorithmic iterations, this encapsulates over 13 million discrete amplitude evaluations.

* **Generated File:** `tlf_dark_matter_bank.h5`
* **Output Specifications:** A Hierarchical Data Format (HDF5) archive containing the full tensor array of probability states. The core dataset, `state_vectors`, is a 2D matrix of shape `[total_steps, 65536]`, storing absolute probabilities encoded as compressed `float32` arrays to optimize storage.
* **Primary Use Case:** HDF5 is the scientific gold standard for large-scale numerical data. Because it supports "lazy loading," researchers can query a highly specific algorithmic slice (e.g., `dark_matter[142, :]`) into memory without requiring massive computational overhead. This serves as the raw mathematical blueprint for rendering exact 3D point clouds, volumetric probability density functions, or further topological analysis in software like **Blender**, **MATLAB**, or **Python (`h5py`)**.

*Executing this cell guarantees the preservation and portability of the raw algorithmic state, providing the foundational architecture required for external, independent exploration.*

````python
# --- 1. INITIALIZE GCP & FETCH METADATA ---
import os
bq_client = bigquery.Client(project=PROJECT_ID)
storage_client = storage.Client(project=PROJECT_ID)
bucket = storage_client.bucket(BUCKET_NAME)

print("📡 Fetching Universe Metadata...")
query = f"SELECT step, epsilon_k, shannon_entropy, state_vector_uri FROM `{AUTO_TABLE_REF}` WHERE event_id = '{auto_event_id}' ORDER BY step ASC"
df_render = bq_client.query(query).to_dataframe()

# Filter active algorithmic window
df_render = df_render[(df_render['shannon_entropy'] < 15.9) & (df_render['shannon_entropy'] > 0.1)].reset_index(drop=True)
total_steps = len(df_render)

# --- 2. DERIVE FUNDAMENTAL PHYSICS CONSTANTS ---
G_TLF = df_render['epsilon_k'].max() / QUBITS
friction_diffs = np.abs(np.diff(df_render['epsilon_k']))
friction_diffs_pos = friction_diffs[friction_diffs > 0]
H_BAR_TLF = np.min(friction_diffs_pos) if len(friction_diffs_pos) > 0 else 1e-4

# --- 3. BUILD THE CAUSAL DAG (LUMINOUS MATTER) ---
print("⚡ Forging the Causal Ledger...")
OPT_GROVER_STEPS = int(np.floor((np.pi / 4.0) * np.sqrt(N_SPACE)))

G = nx.DiGraph()
G.add_node(0, entropy=df_render.iloc[0]['shannon_entropy'], pos=(0.0, 0.0, -1.5), step_created=0, mass=1.0)
nodes_at_level = [0]
node_counter = 1

for idx in range(1, total_steps):
    row = df_render.iloc[idx]
    ent = row['shannon_entropy']

    # Dynamically derive accretion radii and mass limits based on physical constants
    merge_radius = ((QUBITS - ent) / QUBITS) * (G_TLF * np.pi) + (H_BAR_TLF * 10)
    base_child_mass = max(H_BAR_TLF * 10, (QUBITS / (ent + H_BAR_TLF)) * G_TLF)
    num_parent_distinctions = max(1, int((ent / QUBITS) * 4))
    radius_spread = (ent / QUBITS) * 0.8 + 0.1

    new_level_nodes = []
    level_positions = []

    for parent in nodes_at_level:
        px, py, pz = G.nodes[parent]['pos']
        for _ in range(num_parent_distinctions + random.randint(0, 1)):
            angle = random.uniform(0, 2 * np.pi)
            nx_pos = px + radius_spread * np.cos(angle)
            ny_pos = py + radius_spread * np.sin(angle)
            nz_pos = ((QUBITS - ent) / QUBITS) * 3.0 - 1.5 + random.uniform(-0.1, 0.1)

            new_pos = np.array([nx_pos, ny_pos, nz_pos])
            collision = False

            # Collision Detection: Gravitational Accretion
            for e_idx, e_node in enumerate(new_level_nodes):
                if np.linalg.norm(new_pos - level_positions[e_idx]) < merge_radius:
                    G.nodes[e_node]['mass'] += base_child_mass
                    G.add_edge(parent, e_node)
                    collision = True
                    break

            if not collision:
                G.add_node(node_counter, entropy=ent, pos=(nx_pos, ny_pos, nz_pos), step_created=idx, mass=base_child_mass)
                G.add_edge(parent, node_counter)
                new_level_nodes.append(node_counter)
                level_positions.append(new_pos)
                node_counter += 1

    # Apply Pi/4 Grover Limit to surviving nodes
    if len(new_level_nodes) > OPT_GROVER_STEPS:
        new_level_nodes = sorted(new_level_nodes, key=lambda n: G.nodes[n]['mass'], reverse=True)[:OPT_GROVER_STEPS]
    nodes_at_level = new_level_nodes

# --- 4. PREPARE OUTPUT DIRECTORY ---
output_dir = "outputs/data/"
os.makedirs(output_dir, exist_ok=True)

# --- 5. EXPORT LUMINOUS MATTER (CSV) ---
print(f"💾 Exporting Nodes and Edges to {output_dir}...")
node_data = [{'id': n, 'step': d['step_created'], 'entropy': d['entropy'], 'mass': d['mass'],
              'x': d['pos'][0], 'y': d['pos'][1], 'z': d['pos'][2]} for n, d in G.nodes(data=True)]
df_nodes = pd.DataFrame(node_data)
df_nodes.to_csv(f"{output_dir}tlf_universe_nodes.csv", index=False)

edge_data = [{'source': u, 'target': v} for u, v in G.edges()]
df_edges = pd.DataFrame(edge_data)
df_edges.to_csv(f"{output_dir}tlf_universe_edges.csv", index=False)

# --- 6. EXPORT DARK MATTER (HDF5) ---
print(f"🌌 Archiving Dark Matter (Full State Vectors) to {output_dir}...")
with h5py.File(f'{output_dir}tlf_dark_matter_bank.h5', 'w') as hf:

    # Save the physical constants as attributes
    hf.attrs['QUBITS'] = QUBITS
    hf.attrs['G_TLF'] = G_TLF
    hf.attrs['H_BAR_TLF'] = H_BAR_TLF
    hf.attrs['STEPS'] = total_steps

    # Create dataset for optimal float32 compression
    dark_matter = hf.create_dataset('state_vectors', (total_steps, int(N_SPACE)), dtype='float32', compression="gzip")

    for i in range(total_steps):
        print(f"   -> Downloading and compressing step {i}/{total_steps-1}...", end='\r')
        uri = df_render.iloc[i]['state_vector_uri']
        blob_path = uri.replace(f"gs://{BUCKET_NAME}/", "")

        raw_bytes = bucket.blob(blob_path).download_as_bytes()
        complex_vector = np.load(io.BytesIO(raw_bytes))

        # Convert complex amplitudes to absolute probabilities
        probability_vector = np.abs(complex_vector)**2
        dark_matter[i, :] = probability_vector.astype(np.float32)

print("\n🎉 MEMORY BANK COMPILED SUCCESSFULLY!")
print(f"Files saved in: ./{output_dir}")
````

## 🌍 Interactive Holographic Visualization

The following script bypasses the need for external tools by reading the generated `tlf_universe_nodes.csv` and `tlf_universe_edges.csv` files directly back into the notebook. Using Plotly, it renders a fully interactive, 3D holographic map of the causal structures.

````python
# --- 1. LOAD LUMINOUS MATTER ---
print("🌌 Loading Luminous Matter from CSV...")
output_dir = "outputs/data/"

df_nodes = pd.read_csv(f"{output_dir}tlf_universe_nodes.csv")
df_edges = pd.read_csv(f"{output_dir}tlf_universe_edges.csv")

# Index nodes for rapid coordinate lookup
nodes_dict = df_nodes.set_index('id').to_dict('index')

# --- 2. FORGE CAUSAL LINKS (EDGES) ---
print("⚡ Forging Causal Links...")
edge_x, edge_y, edge_z = [], [], []

for _, row in df_edges.iterrows():
    src, tgt = row['source'], row['target']

    if src in nodes_dict and tgt in nodes_dict:
        x0, y0, z0 = nodes_dict[src]['x'], nodes_dict[src]['y'], nodes_dict[src]['z']
        x1, y1, z1 = nodes_dict[tgt]['x'], nodes_dict[tgt]['y'], nodes_dict[tgt]['z']

        # 'None' prevents Plotly from connecting all lines continuously
        edge_x.extend([x0, x1, None])
        edge_y.extend([y0, y1, None])
        edge_z.extend([z0, z1, None])

trace_edges = go.Scatter3d(
    x=edge_x, y=edge_y, z=edge_z,
    mode='lines',
    line=dict(color='rgba(0, 255, 255, 0.3)', width=1.5),
    hoverinfo='none'
)

# --- 3. RENDER SUPERMASSIVE NODES (VERTICES) ---
print("🌍 Rendering Supermassive Nodes...")

# Bound visual sizes to prevent screen eclipsing
scaled_sizes = np.clip(df_nodes['mass'] * 4.0, 2, 50)

# Format analytical hover data
hover_text = [
    f"<b>Node ID:</b> {row['id']}<br>"
    f"<b>Step Born:</b> {row['step']}<br>"
    f"<b>Mass:</b> {row['mass']:.2f}<br>"
    f"<b>Entropy:</b> {row['entropy']:.4f}"
    for _, row in df_nodes.iterrows()
]

trace_nodes = go.Scatter3d(
    x=df_nodes['x'], y=df_nodes['y'], z=df_nodes['z'],
    mode='markers',
    marker=dict(
        size=scaled_sizes,
        color=df_nodes['step'],
        colorscale='Magma',
        opacity=0.9,
        line=dict(width=0),
        colorbar=dict(title="Time (Step Born)", x=0.9)
    ),
    text=hover_text,
    hoverinfo='text'
)

# --- 4. ASSEMBLE THE HILBERT SPACE ---
layout = go.Layout(
    title="Interactive TLF Causal DAG (Luminous Matter)",
    titlefont=dict(color='#00ffff', family='monospace'),
    scene=dict(
        xaxis=dict(showbackground=False, visible=False),
        yaxis=dict(showbackground=False, visible=False),
        zaxis=dict(showbackground=False, visible=False),
        bgcolor='black'
    ),
    paper_bgcolor='black',
    margin=dict(l=0, r=0, b=0, t=40)
)

fig = go.Figure(data=[trace_edges, trace_nodes], layout=layout)

print("✨ Visualization Ready! Click and drag to explore.")
fig.show()
````
